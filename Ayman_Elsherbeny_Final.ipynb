```python
import urllib.request
import os
from pathlib import Path

print("2/3 - تحميل الكود...")
BASE = "https://raw.githubusercontent.com/cherpo1234/ayman-elsherbeny-automation/main"

files = {
    "main.py": "main.py",
    "config/config.yaml": "config/config.yaml",
    "ayman_elsherbeny_automation/__init__.py": "ayman_elsherbeny_automation/__init__.py",
    "ayman_elsherbeny_automation/config.py": "ayman_elsherbeny_automation/config.py",
    "ayman_elsherbeny_automation/core/__init__.py": "ayman_elsherbeny_automation/core/__init__.py",
    "ayman_elsherbeny_automation/core/automation.py": "ayman_elsherbeny_automation/core/automation.py",
    "ayman_elsherbeny_automation/generation/__init__.py": "ayman_elsherbeny_automation/generation/__init__.py",
    "ayman_elsherbeny_automation/generation/video_generator.py": "ayman_elsherbeny_automation/generation/video_generator.py",
    "ayman_elsherbeny_automation/generation/text_to_image.py": "ayman_elsherbeny_automation/generation/text_to_image.py",
    "ayman_elsherbeny_automation/generation/audio_generator.py": "ayman_elsherbeny_automation/generation/audio_generator.py",
    "ayman_elsherbeny_automation/generation/video_merger.py": "ayman_elsherbeny_automation/generation/video_merger.py",
    "ayman_elsherbeny_automation/utils/__init__.py": "ayman_elsherbeny_automation/utils/__init__.py",
    "ayman_elsherbeny_automation/utils/device.py": "ayman_elsherbeny_automation/utils/device.py",
    "ayman_elsherbeny_automation/utils/logging.py": "ayman_elsherbeny_automation/utils/logging.py",
}

root = Path("/content/project")
if root.exists():
    import shutil
    shutil.rmtree(root)

for remote, local in files.items():
    path = root / local
    path.parent.mkdir(parents=True, exist_ok=True)
    url = f"{BASE}/{remote}"
    try:
        urllib.request.urlretrieve(url, path)
        print(f"  ✅ {local}")
    except Exception as e:
        print(f"  ❌ {local}: {e}")

os.chdir(root)
print("✅ تم تحميل الكود")
```
